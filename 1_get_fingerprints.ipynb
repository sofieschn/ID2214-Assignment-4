# Representation B: Morgan Fingerprints (ECFP)

This notebook generates **Morgan Fingerprints** (also known as ECFP - Extended Connectivity Fingerprints) as an alternative molecular representation.

**Goal:** Create binary fingerprint vectors that can be used as an alternative to descriptors for machine learning.

**What are Morgan Fingerprints?**
- High-dimensional binary vectors (typically 1024 or 2048 bits)
- Represent the presence of local molecular substructures
- Commonly used in QSAR (Quantitative Structure-Activity Relationship) and cheminformatics
- Capture molecular patterns that may be important for biological activity


### Import Required Libraries

- **pandas/numpy**: Data manipulation
- **rdkit**: Chemical informatics toolkit for generating fingerprints


```python
import numpy as np
import pandas as pd
from rdkit import Chem
from rdkit.Chem import AllChem

#ignore warnings
import warnings
warnings.filterwarnings('ignore')
```

## Step 1: Load Data

Load the training dataset containing SMILES strings and their activity labels.


```python
training_data = pd.read_csv("training_smiles.csv")
print(f"Loaded {len(training_data)} molecules")
training_data.head()
```

## Step 2: Convert SMILES to Molecule Objects

Convert SMILES strings to RDKit molecule objects so we can generate fingerprints.


```python
training_data["SMILES"] = training_data["SMILES"].astype("string")
training_data["MolFromSmiles"] = training_data["SMILES"].apply(Chem.MolFromSmiles)

# Check for invalid molecules
invalid_molecules = training_data["MolFromSmiles"].isna().sum()
print(f"Valid molecules: {len(training_data) - invalid_molecules}")
print(f"Invalid molecules: {invalid_molecules}")
```

## Step 3: Generate Morgan Fingerprints

Generate Morgan fingerprints for each molecule.

**Parameters:**
- **radius**: Radius of the fingerprint (default 2) - controls the size of local neighborhoods
- **n_bits**: Number of bits in the fingerprint (default 2048) - dimension of the binary vector


```python
# Parameters for Morgan fingerprints
radius = 2  # Radius for Morgan fingerprint (commonly 2 or 3)
n_bits = 2048  # Number of bits in the fingerprint (commonly 1024 or 2048)

def generate_morgan_fingerprint(mol, radius=radius, n_bits=n_bits):
    """
    Generate Morgan fingerprint for a molecule.
    
    Parameters:
    - mol: RDKit molecule object
    - radius: Radius of the fingerprint (default 2)
    - n_bits: Number of bits in the fingerprint (default 2048)
    
    Returns:
    - numpy array of the fingerprint, or None if molecule is invalid
    """
    if mol is None:
        return None
    try:
        fp = AllChem.GetMorganFingerprintAsBitVect(mol, radius=radius, nBits=n_bits)
        return np.array(fp)
    except:
        return None

# Generate fingerprints for all molecules
print("Generating Morgan fingerprints...")
fingerprints = training_data["MolFromSmiles"].apply(
    lambda m: generate_morgan_fingerprint(m, radius=radius, n_bits=n_bits)
)

# Filter out None values (invalid molecules) and convert to numpy array
valid_indices = fingerprints.notna()
fingerprint_matrix = np.stack(fingerprints[valid_indices].values)

print(f"\nFingerprint matrix shape: {fingerprint_matrix.shape}")
print(f"Number of valid molecules: {fingerprint_matrix.shape[0]}")
print(f"Fingerprint dimension: {fingerprint_matrix.shape[1]}")
```

## Step 4: Create Fingerprint DataFrame

Create a DataFrame with fingerprints and corresponding labels for easy use in machine learning.


```python
# Create a DataFrame with fingerprints and corresponding labels
# Only include rows where fingerprints were successfully generated
fingerprint_df = training_data[valid_indices][["INDEX", "ACTIVE"]].copy()

# Add fingerprint columns
for i in range(n_bits):
    fingerprint_df[f"fp_{i}"] = fingerprint_matrix[:, i]

print(f"Fingerprint DataFrame shape: {fingerprint_df.shape}")
print(f"Columns: INDEX, ACTIVE, and {n_bits} fingerprint bits (fp_0 to fp_{n_bits-1})")
fingerprint_df.head()
```

## Step 5: Save Fingerprints

Save the Morgan fingerprint matrix in multiple formats for use in machine learning models.

**Saved files:**
- `training_fingerprints.csv`: DataFrame with INDEX, ACTIVE, and all fingerprint bits
- `training_fingerprints_matrix.npy`: Numpy array of fingerprints (for efficient ML use)
- `training_fingerprints_labels.npy`: Numpy array of labels (ACTIVE column)


```python
# Save fingerprints to CSV
fingerprint_df.to_csv("training_fingerprints.csv", index=False)
print("✓ Morgan fingerprints saved to 'training_fingerprints.csv'")

# Also save as numpy array for direct use in ML models
np.save("training_fingerprints_matrix.npy", fingerprint_matrix)
print("✓ Fingerprint matrix saved to 'training_fingerprints_matrix.npy'")

# Save the corresponding labels separately
labels = fingerprint_df["ACTIVE"].values
np.save("training_fingerprints_labels.npy", labels)
print("✓ Labels saved to 'training_fingerprints_labels.npy'")

print("\nAll files saved successfully!")
```
